```python
n = int(input())
arr = list(map(int, input().split()))
result = map(lambda x : x ** 2, arr)
print(sum(list(result)))
```

```python
n = int(input())
arr = list(map(int, input().split()))
result = filter(lambda x : x % 2 == 0 , arr)
print(len(list(result)))
```

```python
n = int(input())
text = input().split()
result = dict(enumerate(text))
for index, value in result.items():
    print(f"{index}:{value}", end=" ")
```

```python
n = int(input())
arr1 = list(map(int, input().split()))
arr2 = list(map(int, input().split()))

res = map(lambda x: x[0] * x[1], zip(arr1, arr2))
print(sum(list(res)))
```

```python
text = input()
res = filter(lambda x : x == 'a'or x == 'i' or x == 'e' or x == 'o' or x == 'u', text)
if len(list(res)) > 0:
    print("Yes")
else: 
    print("No")
```

```python
n = int(input())
arr = list(map(int, input().split()))
if all(x > 0 for x in arr):
    print("Yes")
else:
    print("No")
```
