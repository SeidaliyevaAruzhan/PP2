```python
n = int(input())
def squares(n):
    for i in range(1, n + 1):
        yield i ** 2
for i in squares(n):
    print(i)
```

```python
n = int(input())
def even(n):
    for i in range(0, n + 1):
        if i % 2 == 0:
            yield i
for i in even(n):
    print(i, end=", " if i < n - 1 else "")
```

```python
def divisibility(n):
    for i in range(0, n + 1):
        if i % 3 == 0 and i % 4 == 0 :
            yield i
    

n = int(input())
for i in divisibility(n):
    print(i, end=" ")
```

```python
def squares(a, b):
    for i in range(a, b + 1):
        yield i ** 2
    

a, b = map(int,input().split())
for i in squares(a, b):
    print(i)
```

```python
def gener(n):
    for i in range(n , 0 - 1, -1):
        yield i

n = int(input())
for i in gener(n):
    print(i)
```

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b


n = int(input())
print(*fibonacci(n), sep=",")
```

```python
class Reverse:
    def __init__(self, s):
        self.s = s
        self.i = len(s) - 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.i < 0:
            raise StopIteration
        ch = self.s[self.i]
        self.i -= 1
        return ch


s = input()
for c in Reverse(s):
    print(c, end="")
```

```python
def primes(n):
    for num in range(2, n + 1):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            yield num


n = int(input())

for p in primes(n):
    print(p, end=" ")
```

```python
def powers_of_two(n):
    for i in range(n + 1):
        yield 2 ** i


n = int(input())

for num in powers_of_two(n):
    print(num, end=" ")
```

```python
def limited_cycle(lst, k):
    for _ in range(k):
        for item in lst:
            yield item


lst = input().split()
k = int(input())

for x in limited_cycle(lst, k):
    print(x, end=" ")
```

```python
import json

def apply_patch(source, patch):
    for key, value in patch.items():
        if value is None:
            source.pop(key, None)
        elif key in source and isinstance(source[key], dict) and isinstance(value, dict):
            apply_patch(source[key], value)
        else:
            source[key] = value
    return source


source = json.loads(input())
patch = json.loads(input())

result = apply_patch(source, patch)

print(json.dumps(result, separators=(',', ':'), sort_keys=True))
```

```python
import json

def find_diff(obj1, obj2, path=""):
    diffs = []
    keys = set(obj1.keys()) | set(obj2.keys())

    for key in keys:
        new_path = f"{path}.{key}" if path else key

        if key not in obj1:
            diffs.append(f"{new_path} : <missing> -> {json.dumps(obj2[key], separators=(',', ':'))}")
        elif key not in obj2:
            diffs.append(f"{new_path} : {json.dumps(obj1[key], separators=(',', ':'))} -> <missing>")
        else:
            v1, v2 = obj1[key], obj2[key]
            if isinstance(v1, dict) and isinstance(v2, dict):
                diffs.extend(find_diff(v1, v2, new_path))
            elif v1 != v2:
                diffs.append(f"{new_path} : {json.dumps(v1, separators=(',', ':'))} -> {json.dumps(v2, separators=(',', ':'))}")

    return diffs


obj1 = json.loads(input())
obj2 = json.loads(input())

differences = sorted(find_diff(obj1, obj2))

if differences:
    print("\n".join(differences))
else:
    print("No differences")
```

```python
import json
import re

data = json.loads(input())
q = int(input())

NOT_FOUND = object()  # special marker

def resolve_query(data, query):
    tokens = re.findall(r'[a-zA-Z_]\w*|\[\d+\]', query)
    current = data

    for token in tokens:
        if token.startswith('['):
            index = int(token[1:-1])
            if isinstance(current, list) and 0 <= index < len(current):
                current = current[index]
            else:
                return NOT_FOUND
        else:
            if isinstance(current, dict) and token in current:
                current = current[token]
            else:
                return NOT_FOUND
    return current


for _ in range(q):
    query = input().strip()
    result = resolve_query(data, query)
    if result is NOT_FOUND:
        print("NOT_FOUND")
    else:
        print(json.dumps(result, separators=(',', ':')))
```

```python
from datetime import datetime, timezone, timedelta
import re

def parse_datetime(s):
    m = re.match(r"(\d{4}-\d{2}-\d{2}) UTC([+-]\d{2}):(\d{2})", s)
    date_str, h, mi = m.groups()
    dt = datetime.strptime(date_str, "%Y-%m-%d")
    offset = timedelta(hours=int(h), minutes=int(mi))
    return dt.replace(tzinfo=timezone(offset))

dt1 = parse_datetime(input())
dt2 = parse_datetime(input())

diff_seconds = abs((dt1 - dt2).total_seconds())
print(int(diff_seconds // 86400))
```

```python
from datetime import datetime, timedelta, timezone

def parse_date_tz(s):
    date_part, tz_part = s.split()
    y, m, d = map(int, date_part.split('-'))
    sign = 1 if tz_part[3] == '+' else -1
    h = int(tz_part[4:6])
    mi = int(tz_part[7:9])
    tz = timezone(timedelta(hours=sign*h, minutes=sign*mi))
    return datetime(y, m, d, tzinfo=tz)

def next_birthday(birth_str, current_str):
    birth = parse_date_tz(birth_str)
    current = parse_date_tz(current_str)
    year = current.year
    month, day = birth.month, birth.day
    if month == 2 and day == 29:
        if not (year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)):
            day = 28
    candidate = datetime(year, month, day, tzinfo=current.tzinfo)
    if candidate.astimezone(timezone.utc) < current.astimezone(timezone.utc):
        year += 1
        month, day = birth.month, birth.day
        if month == 2 and day == 29:
            if not (year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)):
                day = 28
        candidate = datetime(year, month, day, tzinfo=current.tzinfo)
    delta = candidate.astimezone(timezone.utc) - current.astimezone(timezone.utc)
    return int(delta.total_seconds() // 86400)

birth_input = input().strip()
current_input = input().strip()
print(next_birthday(birth_input, current_input))
```

```python
from datetime import datetime, timedelta, timezone

def parse_datetime_with_tz(s):
    date_part, time_part, tz_part = s.split()
    y, m, d = map(int, date_part.split('-'))
    h, mi, sec = map(int, time_part.split(':'))
    sign = 1 if tz_part[3] == '+' else -1
    tz_h = int(tz_part[4:6])
    tz_m = int(tz_part[7:9])
    tz = timezone(timedelta(hours=sign*tz_h, minutes=sign*tz_m))
    return datetime(y, m, d, h, mi, sec, tzinfo=tz)

start = parse_datetime_with_tz(input().strip())
end = parse_datetime_with_tz(input().strip())

duration = int((end.astimezone(timezone.utc) - start.astimezone(timezone.utc)).total_seconds())
print(duration)
```

```python
import math

r = float(input())
x1, y1 = map(float, input().split())
x2, y2 = map(float, input().split())

dx = x2 - x1
dy = y2 - y1
a = dx*dx + dy*dy
b = 2*(x1*dx + y1*dy)
c = x1*x1 + y1*y1 - r*r

discriminant = b*b - 4*a*c

if discriminant <= 0:
    length = math.sqrt(a) if c <= 0 else 0.0
else:
    sqrt_disc = math.sqrt(discriminant)
    t1 = (-b - sqrt_disc) / (2*a)
    t2 = (-b + sqrt_disc) / (2*a)
    t_in = max(0, min(t1, t2))
    t_out = min(1, max(t1, t2))
    length = (t_out - t_in) * math.sqrt(a) if t_out >= t_in else 0.0

print(f"{length:.10f}")
```

```python
import math

r = float(input())
x1, y1 = map(float, input().split())
x2, y2 = map(float, input().split())

dx = x2 - x1
dy = y2 - y1
d = math.hypot(dx, dy)

def intersects_circle(x1, y1, x2, y2, r):
    a = x2 - x1
    b = y2 - y1
    t = -(a*x1 + b*y1) / (a*a + b*b)
    t = max(0, min(1, t))
    px = x1 + a*t
    py = y1 + b*t
    return math.hypot(px, py) < r

if not intersects_circle(x1, y1, x2, y2, r):
    print(f"{d:.10f}")
else:
    def tangent_length(x, y):
        dist = math.hypot(x, y)
        return math.sqrt(dist*dist - r*r)
    
    def tangent_angle(x, y):
        phi = math.atan2(y, x)
        alpha = math.acos(r / math.hypot(x, y))
        return phi - alpha, phi + alpha
    
    ta1, ta2 = tangent_angle(x1, y1)
    tb1, tb2 = tangent_angle(x2, y2)
    
    def path(t1, t2):
        p1x, p1y = r*math.cos(t1), r*math.sin(t1)
        p2x, p2y = r*math.cos(t2), r*math.sin(t2)
        d1 = math.hypot(x1 - p1x, y1 - p1y)
        d2 = math.hypot(x2 - p2x, y2 - p2y)
        angle = abs(t2 - t1)
        if angle > math.pi:
            angle = 2*math.pi - angle
        return d1 + d2 + r*angle
    
    length = min(path(ta1, tb1), path(ta1, tb2), path(ta2, tb1), path(ta2, tb2))
    print(f"{length:.10f}")
```

```python
import importlib

q = int(input())
for _ in range(q):
    module_path, attr = input().split()
    try:
        mod = importlib.import_module(module_path)
    except ModuleNotFoundError:
        print("MODULE_NOT_FOUND")
        continue
    if not hasattr(mod, attr):
        print("ATTRIBUTE_NOT_FOUND")
    elif callable(getattr(mod, attr)):
        print("CALLABLE")
    else:
        print("VALUE")
```
