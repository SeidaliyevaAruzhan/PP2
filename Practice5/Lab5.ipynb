```python
import re

text = input()

if re.match(r'^[A-Za-z].*[0-9]$', text):
    print("Yes")
else:
    print("No")
```

```python
import re

text = input()

match = re.search(r'\S+@\S+\.\S+', text)

if match:
    print(match.group())
else:
    print("No email")
```

```python
import re

text = input()
pattern = input()
replacement = input()

result = re.sub(re.escape(pattern), replacement, text)
print(result)
```

```python
import re

text = input()

pattern = re.compile(r'^\d+$')

if pattern.match(text):
    print("Match")
else:
    print("No match")
```

```python
import re

text = input()

result = re.sub(r'\d', lambda m: m.group() * 2, text)
print(result)
```

```python
import re

text = input()

match = re.search(r'Name:\s*(.+),\s*Age:\s*(.+)', text)

if match:
    print(match.group(1), match.group(2))
```

```python
import re

text = input()

dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b', text)
print(len(dates))
```

```python
import re

s = input()
pattern = input()

escaped_pattern = re.escape(pattern)

matches = re.findall(escaped_pattern, s)

print(len(matches))
```

```python
import re

text = input()

pattern = re.compile(r'\b\w+\b')

matches = pattern.findall(text)

print(len(matches))
```
